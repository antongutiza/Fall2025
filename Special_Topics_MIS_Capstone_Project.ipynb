```python
# Cell 5 — main IT assistant code (LangChain + Chroma + LangGraph simplified)
# Replace your existing Cell 5 with this cell.
import os
from typing import List, Dict, Optional
from pydantic import BaseModel, Field
from datetime import datetime

# LangChain/OpenAI imports
from langchain_openai import ChatOpenAI
from langchain_community.vectorstores import Chroma
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.prompts import ChatPromptTemplate

# Gradio for UI
import gradio as gr

# Ensure API key present
if "OPENAI_API_KEY" not in os.environ:
    raise RuntimeError("OPENAI_API_KEY not set; set it in Colab (e.g. os.environ['OPENAI_API_KEY']='sk-...')")

# LLM setup (adjust model if you prefer)
llm = ChatOpenAI(model="gpt-3.5-turbo", temperature=0)

# Embeddings setup (local HuggingFace embedding used to avoid external embedding call)
from langchain_community.embeddings import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

# Build Chroma vector store from local PDFs
def build_pdf_vectorstore(pdf_paths: List[str], persist_dir: str = "./chroma_pdf_kb") -> Chroma:
    docs = []
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
    for path in pdf_paths:
        if not os.path.exists(path):
            print(f"[WARN] PDF not found: {path}")
            continue
        loader = PyPDFLoader(path)
        pages = loader.load()
        for p in pages:
            # add basic metadata so our UI can show filename when useful
            p.metadata.setdefault("source", "pdf")
            p.metadata.setdefault("file_name", os.path.basename(path))
        docs.extend(pages)
    if not docs:
        raise RuntimeError("No documents loaded. Put PDFs in the pdfs/ folder.")
    split_docs = text_splitter.split_documents(docs)
    vectorstore = Chroma.from_documents(documents=split_docs, embedding=embeddings, persist_directory=persist_dir)
    return vectorstore

# Local PDF list (ensure these files exist - created by Cell 4)
pdf_paths = [
    "pdfs/vpn_policy.pdf",
    "pdfs/password_resets.pdf",
    "pdfs/onboarding.pdf",
]
print("Building vector store (this may take ~15-60s on first run)...")
chroma = build_pdf_vectorstore(pdf_paths)
print("Vector store ready. You can now query the assistant.")

# --- Data models ---
class GraphState(BaseModel):
    messages: List[Dict] = Field(default_factory=list)
    user_id: Optional[str] = None
    knowledge_answer: Optional[str] = None
    final_answer: Optional[str] = None

# Knowledge prompt wrapper for using the retrieved context + LLM
knowledge_prompt = ChatPromptTemplate.from_template("""
You are an IT knowledge agent. Use ONLY the provided context to answer the user's question.

User question:
{question}

Context:
{context}

Instructions:
- If context contains explicit numbered steps or runbook text, return (1) a brief Answer, (2) a Steps: numbered list, and (3) a short Notes: section.
- For Password Reset, VPN Access, and Onboarding, always return a Steps: numbered list even if the context is short.
- If context is insufficient, say "I don't know; please contact IT" and suggest next steps.
- Keep the answer concise but actionable.
""")

def format_steps_from_docs(docs):
    """
    Helper: attempt to extract text that looks like steps
    Returns a short combined string of doc snippets (for 'More details' mode).
    """
    snippets = []
    for d in docs:
        # prefer a chunk that contains a digit+dot (like '1.'), or the longest chunk otherwise
        content = d.page_content
        if "1." in content or "Step" in content or "Steps" in content or "runbook" in content.lower():
            # pick up the first 800 chars to include likely step area
            snippets.append(content[:800])
        else:
            snippets.append(content[:400])
    return "\n\n".join(snippets)

def knowledge_node(question: str, always_steps_for: List[str] = None, detailed: bool = False) -> str:
    """
    Query Chroma, call LLM with context, and format a structured answer.
    - always_steps_for: list of keywords; if question matches one, we enforce 'Steps' formatting
    - detailed: whether to include the longer retrieved snippets
    """
    always_steps_for = always_steps_for or ["password", "password reset", "vpn", "onboarding"]
    # retrieve
    docs = chroma.similarity_search(question, k=3)

    if not docs:
        return "I don't know the answer. Please contact IT or open a ticket."

    # build context: short (concise) or longer (detailed)
    context = format_steps_from_docs(docs if detailed else docs)
    # craft prompt
    chain = knowledge_prompt | llm
    answer_msg = chain.invoke({"question": question, "context": context})
    answer_text = answer_msg.content.strip()

    # enforce a Steps section for core topics (Option C behavior)
    q_lower = question.lower()
    if any(k in q_lower for k in always_steps_for):
        # If LLM returned steps already, trust it. Otherwise, try to synthesize.
        if "steps" in answer_text.lower() or "1." in answer_text:
            return answer_text
        # fallback: extract numbered lines from context
        fallback_steps = []
        for d in docs:
            lines = d.page_content.splitlines()
            for ln in lines:
                ln_strip = ln.strip()
                # detect lines that look like steps: start with digit+dot or "1)" or "Step"
                if ln_strip.startswith(("1.", "1)", "Step", "1 -", "1 -")):
                    fallback_steps.append(ln_strip)
                    # gather a few
                    if len(fallback_steps) >= 6:
                        break
            if len(fallback_steps) >= 6:
                break
        if fallback_steps:
            steps_text = "\n".join(fallback_steps)
            return f"Answer: See steps below.\n\nSteps:\n{steps_text}\n\nNotes: See the attached policy snippets for details."
        # ultimate fallback: return the LLM output
        return answer_text

    return answer_text

# -----------------------------
# User-facing convenience wrappers
# -----------------------------
def ask_knowledge(user_message: str, detailed: bool = False) -> str:
    """Main wrapper used by handlers and UI."""
    return knowledge_node(user_message, detailed=detailed)

# --- button functions for the UI (concise + more detailed) ---
def vpn_access():
    return ask_knowledge("How do I access the VPN?")

def vpn_access_more():
    return ask_knowledge("How do I access the VPN? Please include step-by-step instructions and examples.", detailed=True)

def password_reset():
    return ask_knowledge("How do I reset my password?")

def password_reset_more():
    return ask_knowledge("How do I reset my password? Provide step-by-step instructions and troubleshooting tips.", detailed=True)

def onboarding_help():
    return ask_knowledge("What are the steps for onboarding a new employee?")

def onboarding_help_more():
    return ask_knowledge("What are the steps for onboarding a new employee? Provide a step-by-step checklist and common pitfalls.", detailed=True)

# -----------------------------
# Gradio UI (buttons + 'More details' that responds relative to last main button)
# -----------------------------
def main_button_click(button_name: str, state: Dict):
    """
    state is the gr.State object (a Python dict persisted by Gradio).
    We'll keep minimal state: state['last_main'] = button label
    """
    if state is None:
        state = {}
    mapping = {
        "VPN Access": vpn_access,
        "Password Reset": password_reset,
        "Onboarding Help": onboarding_help,
    }
    func = mapping.get(button_name)
    if not func:
        text = "Unknown action."
    else:
        text = func()
    state['last_main'] = button_name
    return text, state

def more_details_click(state: Dict):
    if not state or "last_main" not in state:
        return "Please click a main button first (VPN Access, Password Reset, or Onboarding Help).", state
    last = state['last_main']
    more_map = {
        "VPN Access": vpn_access_more,
        "Password Reset": password_reset_more,
        "Onboarding Help": onboarding_help_more,
    }
    func = more_map.get(last)
    if not func:
        return "No detailed follow-up available for that topic.", state
    text = func()
    return text, state

# Build Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("## AI-Powered IT Support Assistant (RAG + Runbooks)")

    # Simple text input where users can also type free text (optional)
    user_input = gr.Textbox(label="Or type a question", lines=2, placeholder="e.g. I can't log in to VPN")

    # Response area
    response = gr.Textbox(label="Assistant Response", lines=12)

    # Persist simple state as a dict
    last_state = gr.State(value={})

    # Main buttons row
    with gr.Row():
        btn_vpn = gr.Button("VPN Access")
        btn_pass = gr.Button("Password Reset")
        btn_onboard = gr.Button("Onboarding Help")
        btn_more = gr.Button("More Details")

    # Connect main buttons (they update state and response)
    btn_vpn.click(fn=lambda s: main_button_click("VPN Access", s),
                  inputs=[last_state], outputs=[response, last_state])
    btn_pass.click(fn=lambda s: main_button_click("Password Reset", s),
                   inputs=[last_state], outputs=[response, last_state])
    btn_onboard.click(fn=lambda s: main_button_click("Onboarding Help", s),
                      inputs=[last_state], outputs=[response, last_state])

    # More details uses last_state
    btn_more.click(fn=more_details_click, inputs=[last_state], outputs=[response, last_state])

    # Optional: let the text input submit freeform queries (RAG)
    submit = gr.Button("Ask")
    submit.click(fn=lambda q, s: (ask_knowledge(q, detailed=False), s),
                 inputs=[user_input, last_state], outputs=[response, last_state])

print("Launching Gradio demo...")
demo.launch(debug=True)
```
